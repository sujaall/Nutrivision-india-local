```python
import torch
import torch.nn as nn
import torchvision.transforms as transforms
from torchvision import datasets, models
from torch.utils.data import DataLoader
import timm

DATA_PATH = "../data/top30_foods"

train_tf = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(30),
    transforms.ColorJitter(
        brightness=0.4, contrast=0.4,
        saturation=0.4, hue=0.1),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485,0.456,0.406],
        [0.229,0.224,0.225])
])

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485,0.456,0.406],
        [0.229,0.224,0.225])
])

full = datasets.ImageFolder(DATA_PATH, transform=train_tf)
class_names = full.classes
num_classes = len(class_names)
print(f"✅ Classes: {num_classes}")
print(f"✅ Total images: {len(full)}")
print(f"✅ Class list: {class_names}")

train_size = int(0.8 * len(full))
val_size   = len(full) - train_size
train_set, val_set = torch.utils.data.random_split(
    full, [train_size, val_size])
val_set.dataset.transform = val_tf

train_loader = DataLoader(
    train_set, batch_size=32, shuffle=True)
val_loader   = DataLoader(
    val_set, batch_size=32, shuffle=False)

print(f"✅ Train: {len(train_set)} | Val: {len(val_set)}")

device    = torch.device('cpu')
criterion = nn.CrossEntropyLoss()

def train_one_model(model, name, epochs=15):
    model     = model.to(device)
    optimizer = torch.optim.Adam(
        model.parameters(), lr=0.0005)
    scheduler = torch.optim.lr_scheduler\
        .CosineAnnealingLR(optimizer, T_max=epochs)
    best_acc  = 0
    save_path = f'../models/{name}.pth'

    print(f"\n{'='*50}")
    print(f"TRAINING: {name}")
    print(f"{'='*50}")

    for epoch in range(epochs):
        model.train()
        running_loss = 0
        for images, labels in train_loader:
            images = images.to(device)
            labels = labels.to(device)
            optimizer.zero_grad()
            loss = criterion(model(images), labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()

        model.eval()
        correct = total = 0
        with torch.no_grad():
            for images, labels in val_loader:
                images = images.to(device)
                labels = labels.to(device)
                _, pred = model(images).max(1)
                correct += pred.eq(labels).sum().item()
                total   += labels.size(0)

        acc = 100 * correct / total
        scheduler.step()
        print(f"Epoch {epoch+1:02d}/{epochs} | "
              f"Loss: {running_loss/len(train_loader):.3f} | "
              f"Val Acc: {acc:.1f}%")

        if acc > best_acc:
            best_acc = acc
            torch.save({
                'model_state': model.state_dict(),
                'class_names': class_names,
                'model_name':  name
            }, save_path)
            print(f"  ✅ Saved! Best: {best_acc:.1f}%")

    print(f"\n{name} COMPLETE — Best: {best_acc:.1f}%")
    return best_acc

# MODEL 1 — EfficientNet B2
print("\nStarting Model 1 of 3...")
m1   = timm.create_model('efficientnet_b2',
        pretrained=True, num_classes=num_classes)
acc1 = train_one_model(m1, 'model1_efficientnet')

# MODEL 2 — MobileNetV3
print("\nStarting Model 2 of 3...")
m2 = models.mobilenet_v3_large(weights='DEFAULT')
m2.classifier[3] = nn.Linear(1280, num_classes)
acc2 = train_one_model(m2, 'model2_mobilenetv3')

# MODEL 3 — ResNet50
print("\nStarting Model 3 of 3...")
m3     = models.resnet50(weights='DEFAULT')
m3.fc  = nn.Linear(2048, num_classes)
acc3   = train_one_model(m3, 'model3_resnet50')

print(f"\n{'='*50}")
print("ALL 3 MODELS TRAINED!")
print(f"Model 1 EfficientNet B2 : {acc1:.1f}%")
print(f"Model 2 MobileNetV3     : {acc2:.1f}%")
print(f"Model 3 ResNet50        : {acc3:.1f}%")
estimated = (acc1 + acc2 + acc3) / 3
print(f"Expected Ensemble Acc   : ~{estimated:.1f}%+")
print(f"{'='*50}")
print("Now create ensemble_predict.py!")
```
